```python
#Create a csv with no header and 5 columns. Load it with your own column names and makes the first column the index.

import pandas as pd

# 1. Create CSV file without header
data = """101,Namit,19,CSE,85
102,Parth,20,AIML,90
103,Naman,19,CSE,78
104,Pranav,21,IT,88"""

with open("student.csv", "w") as f:
    f.write(data)

# 2. Load CSV with your own column names
df = pd.read_csv(
    "student.csv",
    header=None,
    names=["ID", "Name", "Age", "Department", "Marks"],
    index_col=0
)

print(df)
```

```python
#Write a file using ; as the deliiter and -999 for missing values. Read it so -999 becomes NaN

# 1. Write a file using ; as delimiter
data = """Name;Age;Marks
Namit;19;85
Parth;20;-999
Pranav;-999;78"""

with open("student.csv", "w") as f:
    f.write(data)

# 2. Read file and convert -999 into NaN
df = pd.read_csv(
    "student.csv",
    sep=";",
    na_values=-999
)

print(df)
```

```python
#Export a dataframe with orient="records", read it back with pd.read_json and check equality with .equals

# 1. Create a DataFrame
df = pd.DataFrame({
    "Name": ["Namit", "Parth", "Pranav"],
    "Age": [19, 20, 19],
    "Marks": [85, 90, 78]
})

# 2. Export DataFrame to JSON
df.to_json("student.json", orient="records")

# 3. Read JSON back into a DataFrame
df2 = pd.read_json("student.json", orient="records")

# 4. Check equality
print(df2)
print("Are both DataFrames equal?", df.equals(df2))
```

```python
#Read a big CSV in chunks of 2000 rows and compute the mean of one column per key across all chunks

# 1. File path
file_path = "/content/Titanic-Dataset.csv"

# 2. Initialize dictionaries
total_sum = {}
total_count = {}

# 3. Read CSV in chunks of 2000 rows
for chunk in pd.read_csv(file_path, chunksize=2000):

    # 4. Calculate sum and count of Age for each Pclass
    grouped = chunk.groupby("Pclass")["Age"].agg(["sum", "count"])

    # 5. Combine results from all chunks
    for key, row in grouped.iterrows():
        total_sum[key] = total_sum.get(key, 0) + row["sum"]
        total_count[key] = total_count.get(key, 0) + row["count"]

# 6. Calculate and display final mean
print("Mean Age by Passenger Class:")

for key in sorted(total_sum):
    if total_count[key] > 0:
        mean_age = total_sum[key] / total_count[key]
        print(f"Pclass {key}: {mean_age:.2f}")
```

```python
import json

# Given JSON text
t = '''{
    "shop": "Fresh Mart",
    "open": true,
    "owner": null,
    "items": [
        {"name": "Rice", "price": 60, "qty": 10},
        {"name": "Oil", "price": 150, "qty": 4},
        {"name": "Tea", "price": 120, "qty": 6}
    ]
}'''

# 1. Load JSON into Python object and print its type
data = json.loads(t)
print(type(data))

# 2. Print name of second item and number of items
print(data["items"][1]["name"])
print(len(data["items"]))

# 3. Print owner value and its Python type
print(data["owner"])
print(type(data["owner"]))

# 4. Convert object back to JSON with indentation 2
print(json.dumps(data, indent=2))
```
